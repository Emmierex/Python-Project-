```python
# --- STAGE 1: Définition des fonctions de conversion ---

def euro_vers_usd(euro):
    """Convertit les Euros en Dollars américains (Taux : 1 € = 1.08 $)"""
    taux = 1.08
    return euro * taux

def usd_vers_euro(usd):
    """Convertit les Dollars américains en Euros"""
    taux = 1.08
    return usd / taux

def euro_vers_mad(euro):
    """Convertit les Euros en Dirhams marocains (Taux : 1 € = 11 MAD)"""
    taux = 11
    return euro * taux

def mad_vers_euro(mad):
    """Convertit les Dirhams marocains en Euros"""
    taux = 11
    return mad / taux

def Ngn_vers_mad(ngn):
    """Convertit les Naria  en Dirhams marocains """
    taux = 148
    return ngn / taux
```

```python


# ==========================================
# --- STAGE 3: Logique avec Boucle Replay ---
# ==========================================

print("Bienvenue dans l'assistant de conversion de monnaie")

# We initialize our loop control variable to "oui"
continuer = "oui"

# The loop runs as long as 'continuer' equals "oui"
while continuer.lower() == "oui":
    print("\nChoisissez la conversion :")
    print("1 - Euro → Dollar")
    print("2 - Dollar → Euro")
    print("3 - Euro → Dirham")
    print("4 - Dirham → Euro")
    print("5 - Naria → Dirham")

    choix = input("\nEntrez le numéro de la conversion : ")

    if choix in ["1", "2", "3", "4", "5"]:


        # --- Displaying the chosen exchange rate dynamically ---
        if choix == "1":
            print("\n Taux appliqué : 1 € = 1.08 $")
        elif choix == "2":
            print("\n Taux appliqué : 1 $ = 0.93 € (1 / 1.08)")
        elif choix == "3":
            print("\n Taux appliqué : 1 € = 11 MAD")
        elif choix == "4":
            print("\n Taux appliqué : 1 MAD = 0.09 € (1 / 11)")
        elif  choix == "5":
            print("\n Taux appliqué : 1 # = 148 MAD")

        try:
            montant = float(input("Entrez le montant à convertir : "))

            if choix == "1":
                resultat = euro_vers_usd(montant)
                print(f"\n👉 {montant:.2f} € = {resultat:.2f} $")
            elif choix == "2":
                resultat = usd_vers_euro(montant)
                print(f"\n👉 {montant:.2f} $ = {resultat:.2f} €")
            elif choix == "3":
                resultat = euro_vers_mad(montant)
                print(f"\n👉 {montant:.2f} € = {resultat:.2f} MAD")
            elif choix == "4":
                resultat = mad_vers_euro(montant)
                print(f"\n👉 {montant:.2f} MAD = {resultat:.2f} €")
            elif choix == "5":
                resultat = Ngn_vers_mad(montant)
                print(f"\n👉 {montant:.2f} Ngn = {resultat:.2f} MAD")

        except ValueError:
            print("Erreur : Veuillez entrer un nombre valide pour le montant.")
    else:
        print("Choix invalide. Veuillez choisir un numéro entre 1 et 4.")

    # This fulfills the final project requirement: asking to repeat
    continuer = input("\nVoulez-vous faire une autre conversion ? (oui/non) : ")

# This code runs only when the loop terminates
print("\nMerci d'avoir utilisé l'assistant de conversion. Au revoir !")
```
